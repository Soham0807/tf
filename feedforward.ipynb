```python
#Step I: Import the libraries
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
```

```python
# Step II : Load the Dataset 
(x_train,y_train),(x_test,y_test) = mnist.load_data()
```

```python
# Step III: Preprocessing

#flattening + normalize
x_train = x_train.reshape(-1,784).astype('float32')/255.0
x_test = x_test.reshape(-1,784).astype('float32')/255.0

# one hot encoding 
y_train = to_categorical(y_train,10)
y_test = to_categorical(y_test,10)
```

```python
# Step IV : Model Configurations
model = tf.keras.Sequential([
    tf.keras.layers.Dense(264,activation='relu',input_shape=(784,)),
    tf.keras.layers.Dense(128,activation='relu'),
    tf.keras.layers.Dense(10,activation='softmax')
])
```

```python
# Step V : Model Compilation
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# Step VI: Train the model
history = model.fit(
    x_train,y_train,epochs=10,batch_size=64)
```

```python
# Step VII : Evaluate The model 
loss,accuracy = model.evaluate(x_test,y_test)
print("Loss : ",loss)
print("Accuracy : ",accuracy)
```

```python
# Step VIII : Testing
predicted = model.predict(x_test[:1])
predicted_class = tf.argmax(predicted[0]).numpy()
print("Predicted Class : ",predicted_class)
print("Actual Class : ",tf.argmax(y_test[0]).numpy())
```

```python
# Step IX : Visualize 
plt.plot(history.history['loss'],label='loss')
plt.plot(history.history['accuracy'],color='red',label='accuracy')
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Training loss Curve")
plt.legend()
plt.grid()
plt.show()
```
