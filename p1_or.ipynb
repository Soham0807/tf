```python
#step I: import libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# step II: Create datasets
x = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,1,1,1])
```

```python
# step III: initializations
weights = np.zeros(2)
bias = 0
lr = 0.1
epochs=10
```

```python
# step IV: activation function
def activation(z):
    if z>=0:
        return 1
    return 0
```

```python
# stev V : Training process
for epoch in range(epochs):
    for i in range(4):
        # formula
        z = np.dot(x[i],weights) + bias
        # activation function
        y_pred = activation(z)
        # calc error
        error = y[i] - y_pred
        # update weights and bias
        weights = weights + lr*error*x[i]
        bias = bias + lr*error
```

```python
# step VI: Testing
for i in range(4):
    z = np.dot(x[i],weights) + bias
    y_pred = activation(z)
    print(x[i],y_pred)
```

```python
# Step VII: Visualize
plt.plot(x[:,0],x[:,1],'bo',label='data points')
test_x = np.linspace(-0.5,1.5,100)
test_y = -(weights[0]*test_x + bias)/weights[1]
plt.plot(test_x,test_y,'r-',label='decision boundary')
plt.xlabel("x1")
plt.ylabel("y1")
plt.legend()
plt.grid()
plt.plot()
```
