```python
# Step I : Import libraries
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.regularizers import l2
```

```python
# Step II : Load the Data
(x_train,y_train),(x_test,y_test) = mnist.load_data()
```

```python
# Step III : Preprocessing 

# Flattening + Normalization
x_train = x_train.reshape(-1,784).astype('float32') / 255.0
x_test = x_test.reshape(-1,784).astype('float32') / 255.0

# One Hot Encoding
y_train = to_categorical(y_train,10)
y_test = to_categorical(y_test,10)
```

```python
# Step IV : Model Conf
model = tf.keras.Sequential([
    tf.keras.layers.Dense(128,activation='relu',input_shape=(784,)),
    tf.keras.layers.Dense(64,activation='relu'),
    tf.keras.layers.Dense(10,activation='softmax')
])
```

```python
# Step V : Model Compilation
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# Step VI : Training Phase
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=64
)
```

```python
# Step VII : Evaluate the model 
loss,accuracy = model.evaluate(x_test,y_test)
print("Loss : ",loss)
print("Accuracy : ",accuracy)
```

```python
# Step VIII : Testing 
predicted = model.predict(x_test[:1])
predicted_class = tf.argmax(predicted[0]).numpy()
print("Predicted Class : ",predicted_class)
print("Actual Class : ",tf.argmax(y_test[0]).numpy())
```

```python
# Step IX : Plot Training and Accuracy
plt.plot(history.history['loss'],color='blue',label='Training Loss')
plt.plot(history.history['accuracy'],color='red',label='Accuracy')
plt.xlabel("Epochs")
plt.title("Loss and Accuracy Curve ")
plt.legend()
plt.grid()
plt.show()
```
