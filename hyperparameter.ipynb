```python
# Step I : Import libraries
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import TensorBoard
```

```python
# Step II : Load the Dataset 
(x_train,y_train),(x_test,y_test) = mnist.load_data()
```

```python
# Step III : PreProcessing

#flattening + normalization
x_train = x_train.reshape(-1,784).astype('float32')/255.0
x_test = x_test.reshape(-1,784).astype('float32')/255.0

#one hot encoding
y_train = to_categorical(y_train,10)
y_test = to_categorical(y_test,10)
```

```python
# Step IV : Experimental values
learning_rates = [0.001,0.01,0.1]
batch_sizes = [32,64,128]
accuracies = []
```

```python
# Step V : Model conf , model compile , training
for lr,bs in zip(learning_rates,batch_sizes):

    # model conf
    model = tf.keras.Sequential([
        tf.keras.layers.Dense(128,activation='relu',input_shape=(784,)),
        tf.keras.layers.Dense(64,activation='relu'),
        tf.keras.layers.Dense(10,activation='softmax')
    ])

    # model Compile
    model.compile(
        optimizer = tf.keras.optimizers.Adam(learning_rate=lr),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    # Tensorboard:
    tensorboard_callback = TensorBoard(log_dir="logs/experiment")
    
    # model train
    history = model.fit(
        x_train,
        y_train,
        epochs=2,
        batch_size=bs,
        callbacks=[tensorboard_callback]
    )

    # model evaluation
    loss,accuracy = model.evaluate(x_test,y_test)
    accuracies.append(accuracy)
    print("Loss : ",loss)
    print("Accuracy : ",accuracy)
```
